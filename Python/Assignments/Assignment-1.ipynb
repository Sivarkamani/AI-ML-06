```python
# Q1. Movie Ticket Counter
```

```python
age = int(input("Enter your age: "))

if age < 5:
    print("Your ticket is Free")
elif age < 18:
    print("Your ticket price is Rs. 100")
elif age < 60:
    print("Your ticket price is Rs. 200")
else:
    print("Your ticket price is Rs. 120")
```

```python
# Q2. Electricity Bill Calculator
```

```python
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 3
elif units <= 200:
    bill = 100 * 3 + (units - 100) * 5
else:
    bill = 100 * 3 + 100 * 5 + (units - 200) * 8

print("Total bill: Rs.", bill)
```

```python
# Q4. Savings Piggy Bank
```

```python
total = 0
days = 0

while total < 1000:
    saved = int(input("Enter today's savings: "))
    total = total + saved
    days = days + 1

print("Goal reached in", days, "days! Total saved: Rs.", total)
```

```python
# Q5. ATM PIN Check
```

```python
correct_pin = 4321
attempts = 0

while attempts < 3:
    pin = int(input("Enter PIN: "))
    if pin == correct_pin:
        print("Access granted")
        break
    else:
        print("Wrong PIN")
        attempts = attempts + 1
else:
    print("Card blocked")
```

```python
# Q6. Even and Odd Counter
```

```python
n = int(input("Enter N: "))
even = 0
odd = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even numbers:", even)
print("Odd numbers:", odd)
```

```python
# Q7. Number Guessing Game
```

```python
secret = 27
guess = 0
count = 0

while guess != secret:
    guess = int(input("Guess the number: "))
    count = count + 1
    if guess > secret:
        print("Too high")
    elif guess < secret:
        print("Too low")

print("Correct! You took", count, "guesses.")
```

```python
# Q8. Exam Result with Grade
```

```python
marks = int(input("Enter marks: "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
elif marks >= 35:
    print("Grade: D")
else:
    print("Fail")
```

```python
# Q9. Sum of Digits on a Bill Number
```

```python
n = int(input("Enter bill number: "))
total = 0

while n > 0:
    digit = n % 10      # take the last digit
    total = total + digit
    n = n // 10         # drop the last digit

print("Sum of digits:", total)

if total > 20:
    print("Congratulations! You win a free gift.")
else:
    print("Sorry, no gift this time.")
```

```python
# Q10. Staircase Pattern
```

```python
steps = int(input("Enter number of steps: "))

for i in range(1, steps + 1):
    for j in range(i):
        print("*", end=" ")
    print()
```
