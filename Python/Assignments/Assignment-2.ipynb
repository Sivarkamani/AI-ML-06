```python
# Q2. Grocery Shopping List
```

```python
groc = []
while(True):
    item= input("Enter item: ")
    if item == "done":
        break;
    elif item in groc:
        print("Item already present, Do not Add")
        
    else:
        groc.append(item)

for i in range(len(groc)):
    print(str(i + 1) + ". " + groc[i])

print("Total items:", len(groc))
       
        
    
```

```python
#Q4. Vowel Counter for a Tweet
```

```python
sentence = input("Enter a sentence: ")
vowels = 0
consonants = 0
spaces = 0

for ch in sentence.lower():
    if ch in "aeiou":
        vowels = vowels + 1
    elif ch.isalpha():
        consonants = consonants + 1
    elif ch == " ":
        spaces = spaces + 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Spaces:", spaces)
```

```python
#Q6. Temperature Tracker
```

```python
temps = [32, 35, 38, 41, 36, 29, 40]
days  = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

print("Heat alert days:")
for i in range(len(temps)):
    if temps[i] > 37:
        print(days[i], "-", temps[i])

hottest = 0
for i in range(len(temps)):
    if temps[i] > temps[hottest]:
        hottest = i
print(f"Hottest day: {days[hottest]} with {temps[hottest]}C")

fahrenheit = []
for t in temps:
    fahrenheit.append(t * 9 / 5 + 32)
print("In Fahrenheit:", fahrenheit)
```

```python
# Q7. Word Frequency in a Sentence
```

```python
sen = input("Enter the sentence") 
words = sen.split()
long = words[0]
short = words[0]
cap = 0
for i in words:
    if len(i) > len(long):
        long = i
    if len(i) < len(short):
        short = i
    if i[0].isupper():
        cap = cap + 1

print("Longest word:", long)
print("Shortest word:", short)
print("Words starting with capital letter:", cap)
```

```python
# Q8. Remove Negative Balances
```

```python
balances = [1500, -200, 3000, -50, 0, 750, -1000]
pos_bal = []
neg_bal = []
for i in balances:
    if i >= 0:
        pos_bal.append(i)
    else:
        neg_bal.append(i)
print(f"Positive balances are: {pos_bal}")
print(f"Negative balances are: {neg_bal}")

amt = sum(neg_bal)
print(f"Total amount owed: {amt}")

```

```python
# Q9. Password Strength Checker
```

```python
password = input("Enter password: ")

has_upper = False
has_lower = False
has_digit = False

for ch in password:
    if ch.isupper():
        has_upper = True
    elif ch.islower():
        has_lower = True
    elif ch.isdigit():
        has_digit = True

met = 0
if has_upper:
    met = met + 1
if has_lower:
    met = met + 1
if has_digit:
    met = met + 1
if len(password) >= 8:
    met = met + 1

if met == 4:
    print("Strong")
elif met >= 2:
    print("Medium")
else:
    print("Weak")
```

```python
# Q10. Simple Name Initials Generator
```

```python
name = input("Enter full name: ")
words = name.split()

initials = ""
for w in words:
    initials = initials + w[0].upper() + "."
print("Initials:", initials)

print("Title case:", name.title())

short = ""
for i in range(len(words) - 1):
    short = short + words[i][0].upper() + ". "
short = short + words[-1].capitalize()
print("Short name:", short)
```
